## Airbnb Data set Description.
## 1. id : unique id for each listing
## 2. host_id : unique id for the host
## 3. host_name : Name of the Host
## 4. Neighbourhood_group : Broad area(e.g., manhattan,Brooklyn)
## 5. neighbourhood : specific locality within the groups.
## 6. latitude & longitude : geographica coordinates.
## 7. name : title of the listing
## 8. room_type
    Types of property:
      # Entire home/apartment
      # private room
      # shared room
## 9. price : cost per night (may contains $,needs cleaning)
## 10. minimum nights : minimun number of nights required to book
## 11. number of reviews : Total review received
## 12. last review : Data of the most recent reviews.
## 13. reviews_per_month : Average number of reviews per month.

## Renaming the columns

```python
df.columns=df.columns.str.lower().str.replace(' ','_')
```

```python
df.columns
```

```python
df['service_fee']
```

```python
df['price']
```

```python
df['price']=df['price'].replace('[$,]','',regex=True).astype(float)
```

```python
df['service_fee']=df['service_fee'].replace('[$,]','',regex=True).astype(float)
```

```python
df['service_fee']
```

## converting the columns into their datatypes

```python
## price and service fee and instant_bookable
col = ['price', 'service_fee']

for i in col:
    df[i] = df[i].astype(float)
```

## check the count of duplicated rows

```python
df.duplicated().sum()
```

## Drop the duplicates

```python
df=df.drop_duplicates()
```

```python
df.duplicated().sum()
```

## checking missing values

```python
df.isnull().sum()
```

## Dropping the columns which has 50% of the data missing

```python
threshold=len(df)*0.5
df=df.dropna(thresh=threshold,axis=1)
```

```python
df.shape
```

```python
num_col=df.select_dtypes(include='number').columns
num_col
```

## Box plot of the Numerical Data

```python
for col in num_col:
  plt.figure(figsize=(5,3))
  sns.boxplot(x=df[col])
  plt.title(f"Box plot of {col} ")
  plt.show()
```

## Filling the Missing column of numerical data with median

```python
df[num_col]=df[num_col].fillna(df[num_col].median())
```

```python
df[num_col].isnull().sum()
```

```python
categorical_col=df.select_dtypes(include='object').columns
categorical_col
```

```python
df[categorical_col].isnull().sum()
```

## Replacing the null values of categorical columns with mode.

```python
for col in categorical_col:
  df[col]=df[col].fillna(df[col].mode()[0])
```

```python
df[categorical_col].isnull().sum()
```

## Formatting the inconsistent data of neighbourhood group.

```python
df['neighbourhood_group']=df['neighbourhood_group'].str.lower()
```

```python
df['neighbourhood_group'].unique()
```

```python
df['neighbourhood_group']=df['neighbourhood_group'].replace({'brookln':'brooklyn','manhatan':'manhattan'})
```

```python
df['neighbourhood_group'].unique()
```

## Removing the outliers from the price column

```python
Q1=df['price'].quantile(0.25)
Q3=df['price'].quantile(0.75)
```

```python
Q1,Q3
```

```python
IQR=Q3-Q1
IQR
```

```python
low_limit=Q1-1.5*IQR
upp_limit=Q3+1.5*IQR
low_limit,upp_limit
```

```python
df = df[(df['price'] > low_limit) & (df['price'] < upp_limit)]
```

```python
df.shape
```

## Boxplot of the price column to check whether there are any outliers

```python
sns.boxplot(x=df['price'],data=df)
plt.show()
```
